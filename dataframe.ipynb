```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name":["Aman","Biswas","Chatur","Depak"],
    "Gender":list("MFFM"),
    "City":["Mumbai","Delhi","Kolkata","Pune"],
    "Salary": [10000,20000,30000,40000]
}

df = pd.DataFrame(data)
df
```

```python
list_of_list_data = [
    ["Aman","M","Mumbai",10000],
    ["Biswas","F","Delhi",20000],
    ["Chatur","F","Kolkata",30000],
    ["Depak","M","Pune",40000]
]

df2 = pd.DataFrame(list_of_list_data)
df2
```

```python
# we need to also give column name seperately
column_name = ["Name","Gender","City","Salary"]
df2 = pd.DataFrame(list_of_list_data,columns=column_name)
df2
```

#### Removing column
- df.drop("ColumnName",axis = 1)
- axis = 1 , means column
- asix = 0 , means row

df.drop("ColumnName",axis = 1)
- 👆 means find "ColumnName" in Column Axis
- if found delete it if not found throw error.
- By default value is axis=0

- There is another parameter inplace = True/False
- True  = Do updates in orignal df
- False = Do updates in a copy of df

```python
df.drop("New_col",axis=1,inplace=True)
df
```
