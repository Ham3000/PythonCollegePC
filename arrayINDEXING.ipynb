#### Slicing 2d matrix

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[3][3] 

# arr[rowIndex][ColIndex] to get an element
# if you only give row it will print whole row

# OR

# arr[rowIndex,ColIndex]
```

```python
arr[3,3]
```

```python
# slicng
# arr[rowStartIndex : rowEndIndex , ColStartIndex : ColEndIndex]
# Note : EndIndex will not be included

arr
```

```python
arr[1:4,0:3]
```
