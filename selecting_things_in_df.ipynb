```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name":["Aman","Biswas","Chatur","Depak"],
    "Gender":list("MFFM"),
    "City":["Mumbai","Delhi","Kolkata","Pune"],
    "Salary": [10000,20000,30000,40000]
}

df = pd.DataFrame(data)
df
```

```python
# .loc[ row , column ] if column not defined then it select all rows
# loc takes lables names . if they are defined then you can pass index
# i am giving index of rows cause i not defined row labels
df.loc[0]
```

```python
# multiple rows
# syntax : df.loc[ [row_index,...] , [column_label,...] ]
df.loc[[0,1]]
```

```python
# just select name and gender
df.loc[[0,1],["Name","Gender"]]
```

#### iloc[row_index,col_index]
- iloc is same as loc but it only takes index as input 
- it does not accept row or column labels

```python
df
```

```python
df.iloc[0] # 0th row and all columns
```

```python
df.iloc[[0,1,2]] # multiple rows all columns
```

```python
df.iloc[:,[1,2]] # all rows and just two columns
```
